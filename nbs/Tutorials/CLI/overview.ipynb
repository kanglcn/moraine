# Tutorials for CLI Overview

This is a series of tutorials to show how to use CLI for InSAR processing. The same five notebooks are given for two data sets (GAMMA results in `data/`):

- [Xinpu](./Xinpu/01_load.ipynb): Sentinel-1 ascending, 60 dates, the Xinpu landslide complex in the Three Gorges Reservoir (vegetated, mostly DS).
- [Campi Flegrei](./CampiFlegrei/01_load.ipynb): Sentinel-1 descending, 92 dates, the uplifting caldera near Naples (urban, many PS).

Run the notebooks of one folder in order; every notebook reads the results of the previous ones in its own folder:

1. `01_load`: GAMMA results to zarr, web mercator coordinates.
2. `02_ps`: PS candidates (amplitude dispersion and Noise2Fringe temporal coherence).
3. `03_ds`: SHP, DS candidates, coherence matrices, phase linking and DS refinement.
4. `04_refine`: merge PS and DS, refine with the Noise2Fringe Transformer.
5. `05_unwrap`: EMCF phase unwrapping and phase closure correction.
